```python
from pathlib import Path

CLIPS = PROJECT / "data" / "clips"

clip_dirs = sorted(p for p in CLIPS.rglob("*") if p.is_dir() and p.name.startswith("SNGS"))
print(len(clip_dirs), "clips found:")
for c in clip_dirs:
    print("  ", c.relative_to(CLIPS))

if clip_dirs:
    c = clip_dirs[0]
    print("\nInside", c.name, ":")
    for p in sorted(c.rglob("*"))[:12]:
        print("  ", p.relative_to(c))
    print(len(list(c.rglob("*.jpg"))), "jpg frames,", len(list(c.rglob("*.mp4"))), "mp4 files")
else:
    print("No SNGS folders found. Showing what's in data/clips:")
    for p in sorted(CLIPS.rglob("*"))[:20]:
        print("  ", p.relative_to(CLIPS))
```

```python
import cv2

VIDEOS = PROJECT / "data" / "videos"
VIDEOS.mkdir(parents=True, exist_ok=True)

def frames_to_video(clip_dir, out_path, fps=25):
    frames_dir = next(Path(clip_dir).rglob("*.jpg")).parent
    frames = sorted(frames_dir.glob("*.jpg"))
    h, w = cv2.imread(str(frames[0])).shape[:2]
    vw = cv2.VideoWriter(str(out_path), cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
    for f in frames:
        vw.write(cv2.imread(str(f)))
    vw.release()
    print(out_path.name, "-", len(frames), "frames,", w, "x", h)

for c in clip_dirs:
    frames_to_video(c, VIDEOS / f"{c.name}.mp4")
```
